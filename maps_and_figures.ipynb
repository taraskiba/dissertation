```python
m = lm.Map(center=(35.5, -82.5), zoom=7)
m.add_gdf(gj, layer_name="COMMONNAME")
legend_dict = {
    "Daniel Boone National Forest": "006400",
    "Chattahoochee-Oconee National Forests": "ffbb22",
    "Cherokee National Forest": "ffff4c",
    "George Washington and Jefferson National Forest": "f096ff",
    "Wayne National Forest": "fa0000",
    "Monongahela National Forest": "b4b4b4",
    "Pisgah National Forest": "f0f0f0",
    "Nantahala National Forest": "0064c8",
    "Shenandoah National Park": "0096a0",
    "Great Smoky Mountains National Park": "00cf75",
}

lm.create_legend(
    title="National Forests and Parks of Southern Appalachia",
    legend_dict=legend_dict
)

m
```
